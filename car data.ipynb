### Data Cleaning: Handling Missing Values

I will fill all missing values (NaN) in the DataFrame with 0. First, let's check the number of missing values per column before cleaning.

```python
print('--- Missing values before filling ---')
display(df.isnull().sum())

# Fill missing values with 0
df_cleaned = df.fillna(0)

print('\n--- Missing values after filling with 0 ---')
display(df_cleaned.isnull().sum())

print('\n--- Displaying head of cleaned DataFrame ---')
display(df_cleaned.head())
```

### Removing Rows with Missing Values (dropna)

```python
print('--- Missing values before mean imputation ---')
display(df.isnull().sum()[df.isnull().sum() > 0])

# Create a copy to avoid modifying the original df directly
df_mean_filled = df.copy()

# Identify numerical columns for mean imputation
numerical_cols = df_mean_filled.select_dtypes(include=['float64', 'int64']).columns

print('\n--- Filling missing values with mean for numerical columns ---')
for col in numerical_cols:
    if df_mean_filled[col].isnull().any():
        mean_val = df_mean_filled[col].mean()
        df_mean_filled[col].fillna(mean_val, inplace=True)
        print(f"Filled missing values in '{col}' with mean: {mean_val:.2f}")

print('\n--- Missing values after mean imputation (numerical columns) ---')
display(df_mean_filled.isnull().sum()[df_mean_filled.isnull().sum() > 0])

print('\n--- Displaying head of DataFrame after mean imputation ---')
display(df_mean_filled.head())
```
